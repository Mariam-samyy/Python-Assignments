```python
##Task-4
##4.1
def calculate_average(scores):
    if len (scores)==0:
        return 0
    else:
        average_scores= sum(scores) / len(scores)
        return average_scores
print(calculate_average([]))

##4.2
def get_result(score, passing_score=60):
    if score < passing_score:
        return "failed"
    else:
        return "passed"
get_result(50)


##4.3
def summarize_student(name, scores):

    average = calculate_average(scores)

    result = get_result(average)

    summary = {
        "name": name,
        "average_score": average,
        "result": result
    }

    return summary

summarize_student("mariam", [90,80,60]) 


##4.4 test
print(calculate_average([]))
print(get_result(60))
print(get_result(65, 70)) ##changing parameter
summarize_student("mariam", [90,80,60])

##4.5
##positional
stu1 = (summarize_student("mariam", [90,80,60]))
print (stu1)

##keyword
stu2= (summarize_student(scores= [30,50,66] ,name = "sarah" ))
print(stu2)
                       
##4.6
def count_results(summaries): 
    passed_count = 0
    failed_count = 0

    for student in summaries:
        if student["result"] == "passed":
            passed_count += 1  
        else:
            failed_count += 1

    return {"passed":passed_count , "failed":failed_count}








                    
                
                      
```

```python
##Task-5
students = [ 
{"name": "Ahmed", "scores": [80, 75, 90]}, 
{"name": "Sara",  "scores": [95, 88, 92]}, 
{"name": "Omar",  "scores": [45, 60, 50]}, 
{"name": "Mona",  "scores": [70, 65, 75]} 
] 

##5.1
students_results = []
for i in students :
    name = i["name"]
    score = i["scores"]
    results = summarize_student(name , score)
    students_results.append(results)

print(students_results)

##5.2
student_rank = sorted(students_results,key=lambda student : student ["average_score"],reverse = True)
print(student_rank)

##5.3
passed_students = list(filter(lambda student : student["result"] == "passed" ,students_results))
print (passed_students)

##5.4
passed_list = list(map(lambda student : student ["name"] ,passed_students))
print(passed_list)

##5.5
for student in student_rank:
    print ( f"name: {student['name']} , average: {student['average_score']:.2f} , result: {student['result']}")

##5.6
results_count = count_results(students_results)
print ("passed:" , results_count["passed"])
print("Failed:" , results_count["failed"])

##5.7
students_alphabetically = sorted(
    students_results,
    key=lambda student: student["name"]
)

print(students_alphabetically)
```

```python
##Task-7
customers = [ 
{"id": 101, "name": " Ahmed ", "city": "cairo"}, 
{"id": 102, "name": "Sara", "city": "GIZA"}, 
{"id": 101, "name": "Ahmed", "city": "Cairo"}, 
{"id": 103, "name": " Omar", "city": "cairo"}, 
{"id": 104, "name": "Mona ", "city": "giza"} 
] 
##7.1
customers_records = []
for i in customers:
    id = i["id"]
    name = i["name"].strip()
    city = i["city"].title()
    new_customer = {"id" : id,
                    "name" : name,
                    "city" : city}
    
    customers_records.append(new_customer)
print(customers_records)

##7.2
unique_customers = []
seen_ids = set()
for i in customers_records:
    id = i["id"]
    if id not in seen_ids:
        seen_ids.add(id)
        unique_customers.append(i)

print(unique_customers)


##7.3
city_count = {}
for i in unique_customers:
    city = i["city"]
    if city in city_count:
        city_count[city] += 1
    else:
        city_count[city] = 1
print (city_count)

##7.4
def clean_customers(records):
    customers_records = []

    for i in records:
        id = i["id"]
        name = i["name"].strip()
        city = i["city"].title()

        new_customer = {
            "id": id,
            "name": name,
            "city": city
        }

        customers_records.append(new_customer)

    unique_customers = []
    seen_ids = set()

    for i in customers_records:
        id = i["id"]

        if id not in seen_ids:
            seen_ids.add(id)
            unique_customers.append(i)

    return unique_customers


print(clean_customers(customers))
print(clean_customers([]))
```

```python
##Task-8
orders = [ 
{"customer": "Ahmed", "amount": 1200, "status": "Completed"}, 
{"customer": "Sara", "amount": 850, "status": "Completed"}, 
{"customer": "Ahmed", "amount": 400, "status": "Cancelled"}, 
{"customer": "Omar", "amount": 600, "status": "Completed"}, 
{"customer": "Sara", "amount": 300, "status": "Completed"} 
]
##8.1
def summarize_orders(orders):
    totals={} 
    for i in orders:
        if i["status"] == "Completed":
            customer = i["customer"]
            amount = i["amount"]

            if customer in totals:
                totals[customer] += amount
            else:
                totals[customer] = amount

    return totals

totals = summarize_orders(orders)
print(totals)

##8.2 totals is defined as an empty dict in the first place

##8.3
ranking = sorted(totals.items(),key = lambda x :x[1] ,reverse = True)
print (ranking)

##8.4
highest_customer = max(totals, key=lambda x: totals[x])

overall_revenue = sum(totals.values())

print("Highest customer:", highest_customer)
print("Overall revenue:", overall_revenue)

##8.5
def get_customers_above(totals, minimum=1000):
    customers = [name for name, amount in totals.items() if amount > minimum]

    return sorted(customers)
print(get_customers_above(totals))
print(get_customers_above(totals,700))

```
